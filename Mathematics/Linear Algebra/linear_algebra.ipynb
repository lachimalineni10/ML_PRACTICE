Exercises on Linear Algebra

```python
import numpy as np

#1. Vector Operations
v1 = np.array([4,5,6])
v2 = np.array([1,2,3])

print(v1+v2)#Add the two vectors.
print(3*v1)#Multiply v1 by a scalar 3.
print(np.dot(v1,v2))#Compute the dot product of v1 and v2.
```

```python
#2. Matrix Operations

a = np.array([[1,2,3],[4,5,6]])
b = np.array([[7,8,9],[10,11,12]])
print("Add matrices A and B:\n",a+b) #Add matrices A and B
print("A Before Transpose:\n",a)
print("Compute the transpose of matrix A:\n",np.transpose(a))#Compute the transpose of matrix A.
print("Multiply matrix A by a scalar 2:\n",2*a)#Multiply matrix A by a scalar 2.
```

Exercise: Compute the Determinant 

```python
#Problem 1: Determinant of a 2×2 Matrix

A = np.array([[5,-2],[3,1]])
print("det of A:",np.linalg.det(A))#Computing det(A)

#Problem 2: Determinant of a 3×3 Matrix

B = np.array([[2,4,3],[1,5,7],[6,8,9]])
print("det of B:",np.linalg.det(B))#Computing det(B)
```

Exercise: Matrix Inverse

```python
#Problem 1:

A = np.array([[2,5],[3,1]])
B = np.linalg.det(A)
if np.isclose(B,0):
    print("A is NOT  invertible")
else:
    A_in = np.linalg.inv(A)
    print("A inverse is:\n", A_in)
    print("identity matrix:\n", A @ A_in)
```

📌 Exercises: Eigenvalues and Eigenvectors

```python
#📝 Problem 1: Find the Eigenvalues and Eigenvectors of a 2×2 Matrix
A = np.array([[3,-1],[2,4]])
eigen_values,eigen_vectors = np.linalg.eig(A)
print("eigen values of A are:\n",eigen_values)
print("eigen vectors of A are:\n",eigen_vectors)
print("eigen values and eigen vectors of A are:\n",np.linalg.eig(A))
print("\n")

#📝 Problem 2: Eigenvalues of a Singular Matrix
B = np.array([[1,2],[2,4]])
eigen_values,eigen_vectors = np.linalg.eig(B)
print("eigen values of B are:\n",eigen_values)
print("eigen vectors of B are:\n",eigen_vectors)
print("eigen values and eigen vectors of B are:\n",np.linalg.eig(B))
print("\n")

#📝 Problem 3: Eigenvalues and Eigenvectors of a 3×3 Matrix
C = np.array([[6,2,1],[0,3,-1],[2,4,3]])
eigen_values,eigen_vectors = np.linalg.eig(C)
print("eigen values of C are:\n",eigen_values)
print("eigen vectors of C are:\n",eigen_vectors)
print("eigen values and eigen vectors of C are:\n",np.linalg.eig(C))
print("\n")
```
