1️⃣ Creating Arrays

```python
import numpy as np

arr1 = np.array([1,2,3,4,5])
print("1D array:",arr1)

arr2 = np.array([[1,2,3,4],[5,6,7,8]])
print("2D array:",arr2)

arr3 = np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
print("3D array:",arr3)

```

2️⃣ Array Properties in NumPy

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])
print("Shape of Array:", arr.shape)
print("size of array:",arr.size)
print("data type of array:",arr.dtype)
print("number of dimensions of array:",arr.ndim)
print("size of Each item:",arr.itemsize,"bytes")
```

3️⃣ Array Initialization in NumPy

```python
zeroes_1d = np.zeros(5,dtype=float)
print("\n1D Array of Zeros:\n", zeroes_1d)

zeroes_2d = np.zeros((2,3),dtype=int)
print("\n2D Array of Zeros:\n", zeroes_2d)

ones_1d = np.ones(10,dtype=int)
print("\n1D Array of ones:\n", ones_1d)

ones_3d = np.ones((3,4,5),dtype=float)
print("\n3D Array of ones:\n", ones_3d)

empty = np.empty((2, 3),dtype=int)
print("Empty Array:\n", empty)

full = np.full((2,2),4,dtype=str)
print("full type Array:\n", full)

identity = np.eye(3,dtype=int)
print("identity type Array:\n", identity)
```
