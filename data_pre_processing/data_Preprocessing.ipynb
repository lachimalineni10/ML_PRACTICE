Data Collection

```python
import pandas as pd
df = pd.read_csv('titanic.csv')
print("First Five Rows:\n\n",df.head())
```

Data Inspection

```python
print("Summary of columns, non-null count, and data types:\n")
print(df.info(),"\n\n")
print("Statistical summary of numeric columns:\n",df.describe(),"\n\n")
print("List all column names:\n",df.columns,"\n\n")
print("Get number of rows and columns:\n",df.shape,"\n\n")
print("Show the data types of each column:\n",df.dtypes,"\n\n")
print("Count missing (null) values per column :\n",df.isnull().sum(),"\n\n")
print("Count of unique values per column :\n",df.nunique(),"\n\n")
```

Data Cleaning

```python
#Handle Missing Values**
print("Missing Values in Each Column Before imputation:\n\n",df.isnull().sum(),"\n\n")
df['Age'].fillna(df['Age'].median(),inplace=True) #Impute Missing Age with Median
df['Embarked'].fillna(df['Embarked'].mode()[0],inplace=True)# Impute Missing Embarked with Mode
df.drop(columns=['Cabin'],inplace=True)#Drop the Cabin column (too many missing values)
print("Missing Values in Each Column After imputation:\n\n",df.isnull().sum(),"\n\n")
```

```python
#Drop Irrelevant Columns
print("Columns Before imputation:\n\n",df.isnull().sum(),"\n\n")
df.drop(columns=['PassengerId', 'Name', 'Ticket'],inplace=True)#These columns have no predictive power 
                                                                # PassengerId → Just an identifier
                                                                # Name → Too unique
                                                                # Ticket → Too random
print("Columns After imputation:\n\n",df.isnull().sum(),"\n\n")
```

```python
# Encode Categorical Columns
#This is called Label Encoding (basic version).
df['Sex'] = df['Sex'].map({'male':0,'female':1})#.map({...}) → replaces text with numbers
                                                # 'male' → 0, 'female' → 1
# One-Hot Encoding 
df = pd.get_dummies(df,columns=['Embarked'],drop_first=True) 
print("First Five Rows:\n\n",df.head())                                               
print(df.info())                                            
```
